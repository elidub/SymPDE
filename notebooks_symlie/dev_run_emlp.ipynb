```python
x, y = trainset[0]
repin, repout = trainset.rep_in, trainset.rep_out
group = O(5)
# repin, repout = repin(group), repout(group)
# repin, repout = 2*V**0, 2*V**0
print(x.size, y.size)
print(repin, repout)
# vis(repin, repout)

net = emlp.nn.pytorch.EMLP(
# net = emlp.nn.pytorch.MLP(
    rep_in=repin,
    rep_out=repout,
    group=group,
    ch = 3,
    num_layers=3,
)

batch_size = 32
lr = 1e-3
max_epochs = 200

optimizer = torch.optim.Adam(net.parameters(),lr=lr)

def loss(x, y):
    x, y = x.float(), y.float()
    yhat = net(x)
    return ((yhat-y)**2).mean()

def train_op(x, y):
    optimizer.zero_grad()
    lossval = loss(x,y)
    lossval.backward()
    optimizer.step()
    return lossval

trainloader = DataLoader(trainset,batch_size=batch_size,shuffle=True)
testloader = DataLoader(testset,batch_size=batch_size,shuffle=True)

test_losses = []
train_losses = []
for epoch in tqdm(range(max_epochs)):
    train_losses.append(np.mean([train_op(*mb).cpu().data.numpy() for mb in trainloader]))
    if not epoch%10:
        with torch.no_grad():
            test_losses.append(np.mean([loss(*mb).cpu().data.numpy() for mb in testloader]))

plt.plot(np.arange(max_epochs),train_losses,label='Train loss')
plt.plot(np.arange(0,max_epochs,10),test_losses,label='Test loss')
plt.legend()
plt.yscale('log')
```
