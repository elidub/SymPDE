```python
class BinaryLinear(nn.Module):
    def __init__(self, input_features, output_features):
        super(BinaryLinear, self).__init__()
        # Initialize the weights as a Parameter
        self.weights = nn.Parameter(torch.randn(output_features, input_features))

    def forward(self, x):
        # Binarize the weights to 0 or 1
        binary_weights = torch.where(self.weights > 0, torch.ones_like(self.weights), torch.zeros_like(self.weights))
        # Perform the matrix multiplication without bias
        return F.linear(x, binary_weights)
    
bl = BinaryLinear(2, 3)

bl(x)
```
